Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;
using System.Linq;

//Определение делегата длч событий
public delegate void StudentEventHandler(string message);

// 1. БАЗОВЫЙ КЛАСС Student (с новыми атрибутами и событиями)
public class Student
{
    // Инкапсуляция и базовые атрибуты
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }

    // Новые атрибуты (3-4 шт.)
    public int Id { get; set; }
    public string Email { get; set; }
    public double AverageScore { get; set; }
    public bool IsActive { get; set; }

    // Событие для отслеживания изменений статуса студент
    public event StudentEventHandler StatusChanged;

    public Student(string name, int age, int course, int id, string email, double averageScore)
    {
        Name = name;
        Age = age;
        Course = course;
        Id = id;
        Email = email;
        AverageScore = averageScore;
        IsActive = true;
    }

    //методы
    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        StatusChanged?.Invoke($"[СОБЫТИЕ] Студент {Name} обновил email на: {Email}");
    }

    public virtual void PromoteCourse()
    {
        Course++;
        StatusChanged?.Invoke($"[СОБЫТИЕ] Студент {Name} успешно переведен на {Course} курс.");
    }

    public virtual void PassExam(double score)
    {
        AverageScore = (AverageScore + score) / 2;
        Console.WriteLine($"{Name} сдал экзамен с оценкой {score}. Обновленный средний балл: {AverageScore:F2}");
    }

    // Виртуальные.м (полиморфизм)
    public virtual string GetInfo()
    {
        return $"ID: {Id}, Имя: {Name}, Возраст: {Age}, Курс: {Course}, Email: {Email}, Средний балл: {AverageScore:F2}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }
}

// 2. Производный клас
public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    
    // Новые атрибуты бакалавра
    public string Faculty { get; set; }
    public bool HasScholarship { get; set; }
    public int PracticumHours { get; set; }

    public BachelorStudent(string name, int age, int course, int id, string email, double avgScore, string specialty, string faculty, bool hasScholarship, int practicumHours) 
        : base(name, age, course, id, email, avgScore)
    {
        Specialty = specialty;
        Faculty = faculty;
        HasScholarship = hasScholarship;
        PracticumHours = practicumHours;
    }

    //методы бакалавра
    public void CompletePracticum(int hours)
    {
        PracticumHours += hours;
        Console.WriteLine($"Бакалавр {Name} прошел производственную практику (+{hours} ч., всего: {PracticumHours} ч.).");
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе бакалавриата по специальности \"{Specialty}\" (Факультет: {Faculty}).");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}, Факультет: {Faculty}, Стипендия: {(HasScholarship ? "Есть" : "Нет")}";
    }
}

// 3 Производный класс
public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }
    
    //атрибуты магистра
    public int PublishedPapers { get; set; }
    public string LabName { get; set; }
    public bool HasGrant { get; set; }

    public MasterStudent(string name, int age, int course, int id, string email, double avgScore, string advisor, int papers, string labName, bool hasGrant) 
        : base(name, age, course, id, email, avgScore)
    {
        ScientificAdvisor = advisor;
        PublishedPapers = papers;
        LabName = labName;
        HasGrant = hasGrant;
    }
      //методы магистра
    public void PublishPaper(string paperTitle)
    {
        PublishedPapers++;
        Console.WriteLine($"Магистрант {Name} опубликовал статью: \"{paperTitle}\" (Всего статей: {PublishedPapers}).");
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает магистерский экзамен под руководством {ScientificAdvisor} (Лаборатория: {LabName}).");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}, Лаборатория: {LabName}, Статей: {PublishedPapers}";
    }
}

// 4 Производный класс
public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }
    
    // Новые атрибуты аспиранта
    public int ConferencesCount { get; set; }
    public int TeachingHours { get; set; }
    public bool CandidateExamPassed { get; set; }

    public PhDStudent(string name, int age, int course, int id, string email, double avgScore, string topic, int confCount, int teachingHours, bool examPassed) 
        : base(name, age, course, id, email, avgScore)
    {
        DissertationTopic = topic;
        ConferencesCount = confCount;
        TeachingHours = teachingHours;
        CandidateExamPassed = examPassed;
    }

    //методы аспиранта
    public void TeachStudents(int hours)
    {
        TeachingHours += hours;
        Console.WriteLine($"Аспирант {Name} провел практические занятия (+{hours} ч., всего преподавания: {TeachingHours} ч.).");
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} работает над диссертацией \"{DissertationTopic}\" (Конференций: {ConferencesCount}, Кандидатский экзамен: {(CandidateExamPassed ? "Сдан" : "Не сдан")}).");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {DissertationTopic}, Часов преподавания: {TeachingHours}";
    }
}

//5. ДЕМОНСТРАЦИЯ РАБОТЫ КОЛЛЕКЦИЙ, СОБЫТИЙ И ЛЯМБДА-ВЫРАЖЕНИЙ

Console.WriteLine("=== ИНФОРМАЦИЯ О СТУДЕНТАХ (КОЛЛЕКЦИЯ LIST<T>) ===\n");

// Создаем список List<Student>
List<Student> studentsList = new List<Student>
{
    new BachelorStudent("Иван Петров", 20, 3, 101, "ivan@uni.com", 4.3, "Программная инженерия", "ИВТ", true, 20),
    new MasterStudent("Анна Сидорова", 24, 1, 102, "anna@uni.com", 4.8, "Проф. Смирнов", 2, "AI Lab", true),
    new PhDStudent("Дмитрий Иванов", 28, 2, 103, "dmitry@uni.com", 4.9, "Искусственный интеллект в образовании", 5, 48, true)
};

// Подписываемся на события изменения статуса каждого студента с помощью лямбда-выражений
foreach (var student in studentsList)
{
    student.StatusChanged += message => Console.WriteLine(message);
}

foreach (var student in studentsList)
{
    Console.WriteLine("--- " + student.GetType().Name + " ---");
    Console.WriteLine(student.GetInfo());
    student.Study();
    student.TakeExam();
    Console.WriteLine();
}

// Демонстрация работы
Console.WriteLine("=== ТЕСТИРОВАНИЕ СОБЫТИЙ И НОВЫХ МЕТОДОВ ==-\n");
studentsList[0].PromoteCourse(); // Переводим Ивана на следующий курс
studentsList[1].UpdateEmail("anna.new@uni.com"); // Меняем почту Анне
Console.WriteLine();

//словарь Dictionary<int, Student> по ID студента
Console.WriteLine("=== СЛОВАРЬ (DICTIONARY<int, Student>) ===\n");
Dictionary<int, Student> studentsDict = new Dictionary<int, Student>
{
    { 201, new BachelorStudent("Елена Козлова", 21, 2, 201, "elena@uni.com", 4.5, "Экономика", "ЭФ", false, 10) },
    { 202, new MasterStudent("Олег Новиков", 23, 2, 202, "oleg@uni.com", 4.2, "Доц. Васильева", 1, "Robo Lab", false) },
    { 203, new PhDStudent("Мария Соколова", 26, 3, 203, "maria@uni.com", 5.0, "Нейронные сети", 8, 96, true) }
};

// Поиск элемента в словаре и вызов специфических методов
if (studentsDict.TryGetValue(203, out Student foundStudent))
{
    Console.WriteLine($"Найдено в словаре: {foundStudent.Name}");
    if (foundStudent is PhDStudent phd)
    {
        phd.TeachStudents(12); // Вызов метода аспиранта
    }
}
```
