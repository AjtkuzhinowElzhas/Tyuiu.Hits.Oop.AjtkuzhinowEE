<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями

```
public interface IStudent
{
    string GetInfo();
    void Study();
    void TakeExam();
    double GetAverageGrade();
}

public abstract class Student : IStudent
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }
    public string Faculty { get; set; }
    public double Scholarship { get; set; }

    protected double averageGrade;

    public Student(string name, int age, int course, string faculty, double scholarship)
    {
        Name = name;
        Age = age;
        Course = course;
        Faculty = faculty;
        Scholarship = scholarship;
        averageGrade = 0;
    }

    public abstract string GetInfo();
    public abstract void Study();
    public abstract void TakeExam();
    public abstract double GetAverageGrade();

    public virtual void PayScholarship()
    {
        Console.WriteLine($"{Name} получает стипендию {Scholarship} руб.");
    }

    public virtual void ChangeCourse(int newCourse)
    {
        Course = newCourse;
        Console.WriteLine($"{Name} переведён на {Course} курс.");
    }

    public virtual void SetAverageGrade(double grade)
    {
        averageGrade = grade;
        Console.WriteLine($"{Name} имеет средний балл {averageGrade}.");
    }
}

public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    public int Credits { get; set; }
    public string GroupNumber { get; set; }
    public bool HasDormitory { get; set; }

    public BachelorStudent(string name, int age, int course, string faculty, double scholarship,
                           string specialty, int credits, string groupNumber, bool hasDormitory)
        : base(name, age, course, faculty, scholarship)
    {
        Specialty = specialty;
        Credits = credits;
        GroupNumber = groupNumber;
        HasDormitory = hasDormitory;
    }

    public override string GetInfo() =>
        $"Бакалавр: {Name}, возраст: {Age}, курс: {Course}, факультет: {Faculty}, " +
        $"специальность: {Specialty}, группа: {GroupNumber}, кредиты: {Credits}, " +
        $"общежитие: {(HasDormitory ? "да" : "нет")}";

    public override void Study() =>
        Console.WriteLine($"{Name} учится на {Course} курсе бакалавриата по специальности \"{Specialty}\".");

    public override void TakeExam() =>
        Console.WriteLine($"{Name} сдаёт экзамен по специальности \"{Specialty}\".");

    public override double GetAverageGrade() => averageGrade;

    public void ApplyForDormitory()
    {
        HasDormitory = true;
        Console.WriteLine($"{Name} подал заявку на общежитие и получил место.");
    }

    public void AddCredits(int amount)
    {
        Credits += amount;
        Console.WriteLine($"{Name} получил {amount} кредитов. Всего: {Credits}.");
    }
}

public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }
    public string ThesisTopic { get; set; }
    public int Publications { get; set; }
    public bool HasGrants { get; set; }

    public MasterStudent(string name, int age, int course, string faculty, double scholarship,
                         string scientificAdvisor, string thesisTopic, int publications, bool hasGrants)
        : base(name, age, course, faculty, scholarship)
    {
        ScientificAdvisor = scientificAdvisor;
        ThesisTopic = thesisTopic;
        Publications = publications;
        HasGrants = hasGrants;
    }

    public override string GetInfo() =>
        $"Магистр: {Name}, возраст: {Age}, курс: {Course}, факультет: {Faculty}, " +
        $"научный руководитель: {ScientificAdvisor}, тема: \"{ThesisTopic}\", " +
        $"публикаций: {Publications}, гранты: {(HasGrants ? "да" : "нет")}";

    public override void Study() =>
        Console.WriteLine($"{Name} учится в магистратуре и работает над темой \"{ThesisTopic}\".");

    public override void TakeExam() =>
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}.");

    public override double GetAverageGrade() => averageGrade;

    public void PublishArticle()
    {
        Publications++;
        Console.WriteLine($"{Name} опубликовал статью. Всего публикаций: {Publications}.");
    }

    public void ApplyForGrant()
    {
        HasGrants = true;
        Console.WriteLine($"{Name} получил грант на исследование.");
    }
}

public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }
    public string ResearchArea { get; set; }
    public int YearsOfStudy { get; set; }
    public bool HasDefense { get; set; }

    public PhDStudent(string name, int age, int course, string faculty, double scholarship,
                      string dissertationTopic, string researchArea, int yearsOfStudy, bool hasDefense)
        : base(name, age, course, faculty, scholarship)
    {
        DissertationTopic = dissertationTopic;
        ResearchArea = researchArea;
        YearsOfStudy = yearsOfStudy;
        HasDefense = hasDefense;
    }

    public override string GetInfo() =>
        $"Аспирант: {Name}, возраст: {Age}, курс: {Course}, факультет: {Faculty}, " +
        $"тема диссертации: \"{DissertationTopic}\", область: {ResearchArea}, " +
        $"лет обучения: {YearsOfStudy}, защита: {(HasDefense ? "была" : "нет")}";

    public override void Study() =>
        Console.WriteLine($"{Name} работает над диссертацией \"{DissertationTopic}\" в области {ResearchArea}.");

    public override void TakeExam() =>
        Console.WriteLine($"{Name} сдаёт кандидатский экзамен по области {ResearchArea}.");

    public override double GetAverageGrade() => averageGrade;

    public void DefendDissertation()
    {
        HasDefense = true;
        Console.WriteLine($"{Name} успешно защитил диссертацию!");
    }

    public void AttendConference(string conference)
    {
        Console.WriteLine($"{Name} выступил на конференции \"{conference}\".");
    }
}

public class StudentService
{
    private readonly IStudent _student;

    public StudentService(IStudent student)
    {
        _student = student;
    }

    public void ProcessStudent()
    {
        Console.WriteLine(_student.GetInfo());
        _student.Study();
        _student.TakeExam();
        Console.WriteLine($"Средний балл: {_student.GetAverageGrade()}");
        Console.WriteLine();
    }
}

var bachelor = new BachelorStudent("Иван Петров", 20, 3, "ИТ", 3000,
                                   "Программная инженерия", 90, "ПИ-301", false);
var master = new MasterStudent("Анна Сидорова", 24, 1, "ИТ", 5000,
                               "Проф. Смирнов", "Машинное обучение", 3, false);
var phd = new PhDStudent("Дмитрий Иванов", 28, 2, "ИТ", 7000,
                         "Искусственный интеллект в образовании", "ИИ", 2, false);

bachelor.SetAverageGrade(4.5);
master.SetAverageGrade(4.8);
phd.SetAverageGrade(5.0);

bachelor.ApplyForDormitory();
bachelor.AddCredits(10);
master.PublishArticle();
master.ApplyForGrant();
phd.AttendConference("AI Summit 2026");
phd.DefendDissertation();

Console.WriteLine("\n=== СЕРВИС ОБРАБОТКИ СТУДЕНТОВ ===\n");

var services = new List<StudentService>
{
    new StudentService(bachelor),
    new StudentService(master),
    new StudentService(phd)
};

foreach (var service in services)
{
    service.ProcessStudent();
}

Console.WriteLine("=== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА ===\n");

IStudent explicitStudent = bachelor;
Console.WriteLine(explicitStudent.GetInfo());
explicitStudent.Study();
explicitStudent.TakeExam();
Console.WriteLine($"Средний балл: {explicitStudent.GetAverageGrade()}");
```
