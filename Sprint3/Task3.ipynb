<h2 style="color:DodgerBlue">Описание проекта:</h2>

Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student:
• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
• Методы:
o GetInfo(): метод для получения информации о студенте в виде строки.
o Study(): метод для вывода сообщения о том, что студент учится.
o TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:
1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
переопределен для добавления информации о специальности при изучении
предмета.
2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
атрибуты, такие как Научный руководитель (ScientificAdvisor).
Метод TakeExam() должен быть переопределен для добавления
информации о научном руководителе при сдаче экзамена.
3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
Метод Study() должен быть переопределен для добавления информации о
теме диссертации при изучении предмета.
#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public interface IResearcher
{
    void ConductResearch();
    string ResearchTopic { get; set; }
}

public interface ITeacher
{
    void Teach();
    string Subject { get; set; }
}

public class Student
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }
    public string Email { get; set; }
    public double GPA { get; set; }
    public string City { get; set; }
    public DateTime EnrollmentDate { get; set; }

    public static int TotalStudents { get; private set; } = 0;

    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        Email = "unknown@example.com";
        GPA = 0;
        City = "Не указан";
        EnrollmentDate = DateTime.Now;
        TotalStudents++;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Курс: {Course}, Email: {Email}, GPA: {GPA}, Город: {City}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }

    public void SendEmail(string message)
    {
        Console.WriteLine($"Письмо для {Name} ({Email}): {message}");
    }

    public void UpdateGPA(double newGpa)
    {
        if (newGpa >= 0 && newGpa <= 5)
        {
            GPA = newGpa;
            Console.WriteLine($"{Name}: новый GPA = {GPA}");
        }
    }

    public int GetYearsOfStudy() => Course;

    public void InteractWith(Student other)
    {
        Console.WriteLine($"{Name} общается с {other.Name}.");
        other.Study();
    }

    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Общее количество студентов: {TotalStudents}");
    }

    // ПЕРЕГРУЗКА: SubmitWork с разными параметрами
    public void SubmitWork()
    {
        Console.WriteLine($"{Name} сдал работу без указания деталей.");
    }

    public void SubmitWork(string title)
    {
        Console.WriteLine($"{Name} сдал работу: \"{title}\".");
    }

    public void SubmitWork(string title, int pages)
    {
        Console.WriteLine($"{Name} сдал работу: \"{title}\" ({pages} стр.).");
    }

    public void SubmitWork(string title, int pages, bool isLate)
    {
        Console.WriteLine($"{Name} сдал работу: \"{title}\" ({pages} стр.), {(isLate ? "с опозданием" : "в срок")}.");
    }
}

public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    public string Faculty { get; set; }
    public int Credits { get; set; }
    public bool HasScholarship { get; set; }

    public BachelorStudent(string name, int age, int course, string specialty)
        : base(name, age, course)
    {
        Specialty = specialty;
        Faculty = "Не указан";
        Credits = 0;
        HasScholarship = false;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе бакалавриата по специальности \"{Specialty}\".");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}, Факультет: {Faculty}, Кредиты: {Credits}, Стипендия: {(HasScholarship ? "есть" : "нет")}";
    }

    public void ApplyForScholarship()
    {
        HasScholarship = GPA >= 4.5;
        Console.WriteLine($"{Name}: стипендия {(HasScholarship ? "получена" : "не получена")}");
    }

    public void AddCredits(int amount)
    {
        Credits += amount;
        Console.WriteLine($"{Name} получил {amount} кредитов. Всего: {Credits}");
    }

    // ПЕРЕГРУЗКА в наследнике — своя версия SubmitWork с bool
    public void SubmitWork(string title, bool isProject)
    {
        Console.WriteLine($"{Name} сдал {(isProject ? "проект" : "работу")}: \"{title}\".");
    }
}

public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }
    public string ThesisTopic { get; set; }
    public int Publications { get; set; }
    public bool HasInternship { get; set; }

    public MasterStudent(string name, int age, int course, string scientificAdvisor)
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
        ThesisTopic = "Не определена";
        Publications = 0;
        HasInternship = false;
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен под руководством научного руководителя {ScientificAdvisor}.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}, Тема: {ThesisTopic}, Публикаций: {Publications}";
    }

    public void PublishArticle(string title)
    {
        Publications++;
        Console.WriteLine($"{Name} опубликовал статью: \"{title}\". Всего: {Publications}");
    }

    public void StartInternship(string company)
    {
        HasInternship = true;
        Console.WriteLine($"{Name} проходит стажировку в {company}.");
    }
}

public class MasterResearcherStudent : MasterStudent
{
    public string ResearchLab { get; set; }
    public int GrantAmount { get; set; }
    public string ConferenceName { get; set; }

    public MasterResearcherStudent(string name, int age, int course, string advisor, string researchLab)
        : base(name, age, course, advisor)
    {
        ResearchLab = researchLab;
        GrantAmount = 0;
        ConferenceName = "Не указана";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} (магистрант-исследователь) работает в лаборатории \"{ResearchLab}\" над темой \"{ThesisTopic}\".");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Лаборатория: {ResearchLab}, Грант: {GrantAmount} руб., Конференция: {ConferenceName}";
    }

    public void ReceiveGrant(int amount)
    {
        GrantAmount += amount;
        Console.WriteLine($"{Name} получил грант на {amount} руб. Всего: {GrantAmount}");
    }

    public void PresentAtConference(string conference)
    {
        ConferenceName = conference;
        Console.WriteLine($"{Name} выступает на конференции \"{conference}\".");
    }
}

public class PhDStudent : Student, IResearcher, ITeacher
{
    public string DissertationTopic { get; set; }
    public int PublishedPapers { get; set; }
    public int TeachingHours { get; set; }
    public string ResearchTopic { get; set; }
    public string Subject { get; set; }

    public PhDStudent(string name, int age, int course, string dissertationTopic)
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
        PublishedPapers = 0;
        TeachingHours = 0;
        ResearchTopic = dissertationTopic;
        Subject = "Не указан";
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} работает над диссертацией на тему \"{DissertationTopic}\" и учится на {Course} курсе аспирантуры.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: {DissertationTopic}, Публикаций: {PublishedPapers}, Предмет: {Subject}";
    }

    public void ConductResearch() => Console.WriteLine($"{Name} проводит исследование по теме \"{ResearchTopic}\".");
    public void Teach() => Console.WriteLine($"{Name} преподает предмет \"{Subject}\" ({TeachingHours} ч.).");

    public void PublishPaper(string title)
    {
        PublishedPapers++;
        Console.WriteLine($"{Name} опубликовал научную работу: \"{title}\". Всего: {PublishedPapers}");
    }
}

// GENERIC КЛАСС — по образцу VehicleCollection<T>
public class StudentCollection<T> where T : Student
{
    private List<T> _students = new List<T>();

    public void Add(T student)
    {
        _students.Add(student);
    }

    public void Remove(T student)
    {
        _students.Remove(student);
    }

    public void DisplayStudents()
    {
        foreach (var student in _students)
            student.Study();
    }

    public int Count => _students.Count;
}

// ==== ДЕМОНСТРАЦИЯ ====

Console.WriteLine("=== ПОЛИМОРФИЗМ (перекрытие) ===\n");

List<Student> students = new List<Student>
{
    new BachelorStudent("Иван Петров", 20, 3, "Программная инженерия"),
    new MasterStudent("Анна Сидорова", 24, 1, "Проф. Смирнов"),
    new MasterResearcherStudent("Пётр Кузнецов", 25, 2, "Проф. Орлов", "Лаборатория ИИ"),
    new PhDStudent("Дмитрий Иванов", 28, 2, "Искусственный интеллект в образовании")
};

foreach (var student in students)
{
    Console.WriteLine("--- " + student.GetType().Name + " ---");
    Console.WriteLine(student.GetInfo());
    student.Study();
    student.TakeExam();
    Console.WriteLine();
}

Console.WriteLine("=== ПЕРЕГРУЗКА SubmitWork() ===\n");

Student s = new Student("Тестовый Студент", 19, 1);
s.SubmitWork();
s.SubmitWork("Курсовая");
s.SubmitWork("Курсовая", 20);
s.SubmitWork("Курсовая", 20, true);

Console.WriteLine();
BachelorStudent b = new BachelorStudent("Иван Петров", 20, 3, "Программная инженерия");
b.SubmitWork("Дипломный проект", true);

Console.WriteLine("\n=== GENERIC КЛАСС StudentCollection<T> ===\n");

StudentCollection<BachelorStudent> bachelors = new StudentCollection<BachelorStudent>();
bachelors.Add(new BachelorStudent("Елена Козлова", 21, 2, "Экономика"));
bachelors.Add(new BachelorStudent("Олег Новиков", 20, 1, "Менеджмент"));
Console.WriteLine($"Бакалавров: {bachelors.Count}");
bachelors.DisplayStudents();

Console.WriteLine();

StudentCollection<Student> all = new StudentCollection<Student>();
all.Add(new BachelorStudent("Елена Козлова", 21, 2, "Экономика"));
all.Add(new MasterStudent("Олег Новиков", 23, 2, "Доц. Васильева"));
all.Add(new PhDStudent("Мария Соколова", 26, 3, "Нейронные сети"));
Console.WriteLine($"Всего студентов в generic-коллекции: {all.Count}");
all.DisplayStudents();

Console.WriteLine();
Student.DisplayTotalStudents();
```
